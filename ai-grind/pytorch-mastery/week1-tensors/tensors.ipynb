```python
import torch
```

```python
def print_tensor(t):
    print(type(t), t.numel(), t.shape, t.dtype, t)
```

```python
#tensors are DS similar to arrays and matrices. multi dimension array of same type
# tensors are similar to numpy ndarrays

#torch.tensor is a factory method to create tensors.
# creates 0D scalar that contains the value
a = torch.tensor(10)
print("a shape", a.shape)

b = torch.tensor([1, 2])
print("b shape", b.shape)


c = torch.zeros(2, 3)
print(c, c.dtype)

d = torch.zeros([2, 3], dtype=torch.int32)
print(d, d.dtype)

e = torch.zeros([2, 3], dtype=torch.int32, device="mps")
print(e, e.dtype)

ones = torch.ones([2, 3], dtype=torch.int32, device="mps")
print(ones, ones.dtype, ones.shape)

vector = torch.tensor([[[1, 2], [2, 3], [3, 4], [4, 5]], [[2, 3], [3, 4], [4, 5], [5, 6]], [[3, 4], [4, 5], [5, 6], [6, 7]]])
print(vector, vector.dtype, vector.shape)
```

```python
a = torch.arange(15)
print(type(a), a.dtype, a)

b = torch.arange(1, 6)
print(type(b), b.dtype, b)

c = torch.arange(1, 8, 4)
print(type(c), c.dtype, c)

d = torch.arange(1, 7, 4.5)
print(type(d), d.dtype, d)
```

```python
# linspace is a one dimensional tensor evenly spaced by number of steps. 

linspace = torch.linspace(1, 10, 20)
print(type(linspace), linspace.dtype, linspace)

a = torch.linspace(start=1, end=10, steps=20)
print(type(a), a.dtype, a)

b = torch.linspace(-10, 10, 5)
print(type(b), b.dtype, b)

c = torch.linspace(1, 10, 20, device="mps")
print(type(c), c.dtype, b)
```

```python
#numbers from normalized distribution with 0 mean 1 variance.
a = torch.randn(15)
print(type(a), a.dtype, a)

b = torch.randn([3, 4])
print(type(b), b.dtype, b)

c = torch.randn(8, 3, 4, 3)
print(type(c), c.dtype, c)
```

```python
# numbers from uniform distribution between low and high. with size parameter
a = torch.randint(2, 4, (2, 5))
print(type(a), a.numel(), a)

b = torch.randint(2, 4, (3,))
print(type(b), b.numel(), b)
```

```python
x = torch.arange(60)
print_tensor(x)
x = x.reshape(3, 4, 5)
print_tensor(x)
```

```python
# prints first matrix, 0 column index based.
m = x[1]
print_tensor(m)

# prints every 2nd row from each matrix.
m = x[:, 2, :]
print_tensor(m)

# prints every matrix, every row but 0 th element. this flattens each column as a row and returns them
m = x[:, :, 0]
print_tensor(m)

m = x[:, 0, 0]
print_tensor(m)

m = x[:, 1:3, :]
print_tensor(m)

# slicing helps retain dimensions than selecting index.
#x[:, 0, :] and x[:, 0:1, :0]
```

```python
# first 3 columns
m = x[:,:, :3]
print_tensor(m)

# ::n means step by n
m = x[:, :, ::2]
print_tensor(m)

m = x[-1]
print_tensor(m)

m = x[-1, -1, :]
print_tensor(m)
```

```python
m = torch.flip(x, dims=[-1])
print_tensor(m)

n = torch.flip(x, dims=[1])
print_tensor(n)

n = torch.flip(x, dims=[0])
print_tensor(n)

n = torch.flip(x, dims=[2])
print(torch.equal(m, n))
```

```python
#boolean tensors matching elements.
mask = x > 5
print(mask)

# flattened indices of matching mask.
m = x[mask]
print_tensor(m)

# direct masking
m = x[x%2 == 0]
print_tensor(m)

# masking with multiple conditions, &, |, ~

m = x[(x%2 == 0) & (x%3 == 0)]
print_tensor(m)
```

```python
#where condition, failed condition are filled with default value.
m = torch.where(x > 5, x, torch.tensor(0))
print_tensor(m)

m = torch.where(x > 5, x, 0)
print_tensor(m)

m = torch.where(x % 2 == 0, 1, 0)
print_tensor(m)

numbers = torch.tensor([-4, -2, 0, 2, 4])

manual_relu = torch.where(numbers < 0, 0, numbers)
print_tensor(manual_relu)

rel = torch.relu(numbers)
print_tensor(rel)

#indices.
m = torch.index_select(x, dim=0, index=torch.tensor([0, 2]))
print_tensor(m)

m = torch.index_select(x, dim=1, index=torch.tensor([0, 2]))
print_tensor(m)

m = torch.index_select(x, dim=2, index=torch.tensor([0, 2]))
print_tensor(m)

filtered = x[(x >= 20) & (x <= 40)]
print (filtered.min().item() == 20)
print (filtered.max().item() == 40)
```

```python
# crop images.
import matplotlib.pyplot as plt

images = torch.randint(255, [16, 3, 32, 32])
image_tensor = images[0]

# mark first 11 rows red, blue values as 0
# mark first 11 rows green, blue values as 0
# mark first 11 rows red, green values as 0
image_tensor[0,0:11,:]=0
image_tensor[2,0:11,:]=0

image_tensor[1,10:21,:]=0
image_tensor[2,10:21,:]=0
image_tensor[0,21:32,:]=0
image_tensor[1,21:32,:]=0
def print_image(image_tensor):
    image = image_tensor.clone().detach().cpu().permute(1, 2, 0).numpy()
    plt.imshow(image)
    #plt.axis("off")
    plt.show()
print_image(image_tensor)

# Crop the image.
def crop_image(image, crop_height, crop_width):
    _, height, width = image.shape
    start_height = (height - crop_height) // 2
    start_width = (width - crop_width) // 2
    print(start_height,start_width)
    return image[:, start_height: start_height+crop_height, start_width: start_width+crop_width]

c_image = crop_image(image_tensor, 12, 12)
print_image(c_image)


```

```python
a = torch.tensor([1.0, 2.0, 3.0, 4.0], requires_grad=True)
b = a[:2]
s = b.sum()
s.backward()
print_tensor(a)
print_tensor(a.grad)
```
